### Business Insight

This insight can help businesses improve retention strategies

```python
import matplotlib.pyplot as plt
import seaborn as sns 

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Gender")
plt.title ("Customer Segementation Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

### Observation 

-The chart shows the number of customers in each segmentation group.


-This helps identify which customer segment has the highest and lowest number of customers.
